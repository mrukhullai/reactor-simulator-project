# NumValidator

`NumValidator` is a compact helper for checking whether a value is a valid number and, if needed, whether it falls within a specified range. In a notebook, it is useful for validating user input, parameter values, and calculations before they are used.

It typically rejects non-numeric values, booleans, and values outside optional minimum and maximum bounds while keeping the validation logic simple and reusable.

Why not just use `str.isdigit()`?

Because `isdigit()` is only a narrow test for strings made entirely of digits. It returns `False` for values like `3.14`, `-2`, `1e3`, and `' 42 '`, even though these may still be valid numeric inputs. It also does not handle `int`, `float`, `None`, or `bool` in a robust way.

A custom `numvalidator()` function is created to solve that broader problem: it accepts real numeric values, rejects invalid entries such as booleans or non-numeric strings, checks optional minimum/maximum bounds, and works consistently for user input and parameter validation in notebooks or scripts.

```python
def numvalidator(value, *, min_value=None, max_value=None):
    """Return True if value is a real number within optional bounds."""
    if isinstance(value, bool):
        return False

    try:
        number = float(value)
    except (TypeError, ValueError):
        return False

    if number != number or abs(number) == float('inf'):
        return False

    if min_value is not None and number < float(min_value):
        return False

    if max_value is not None and number > float(max_value):
        return False

    return True

# Example usage
samples = [3, "5.2", None, "abc", 42, -1, True]
for sample in samples:
    print(f"{sample!r}: {numvalidator(sample, min_value=0, max_value=10)}")
```
